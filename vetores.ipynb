```python
temperaturas = [28, 30, 27, 31, 29, 26, 25]
```

```python
for pos in range(len(temperaturas)):
  print(f"Dia {pos}: {temperaturas[pos]}°C")
```

```python
maior = temperaturas[0]

for pos in range(1, len(temperaturas)):
  if temperaturas[pos] > maior:
    maior = temperaturas[pos]

print(f"Maior temperatura: {maior}°C")
```
