```python
TAMANHO = 7
codigos = [10, 17, 24, 8, 15]

for codigo in codigos:
  print(f"{codigo} % {TAMANHO} : {codigo % TAMANHO}")
```

```python
tabela = [[] for _ in range(TAMANHO)]
print(tabela)
print("Quantidade de posições: ", len(tabela))
```

```python
produto = {"codigo" : 10, "nome" : "Teclado"}
indice = produto["codigo"] % TAMANHO
tabela[indice].append(produto)

for posicao, itens in enumerate(tabela):
  print(f"{posicao} : {itens}")
```

```python
def inserir(tabela, codigo, nome):
  indice = codigo % TAMANHO

  for produto in tabela[indice]:
    if produto["codigo"] == codigo:
      return False

  tabela[indice].append({"codigo" : codigo, "nome" :nome})
  return True

inserir(tabela, 17, "Mouse")

for posicao, itens in enumerate(tabela):
  print(f"{posicao} : {itens}")
```

```python
inserir(tabela, 8, "Mouse")
for posicao, itens in enumerate(tabela):
  print(f"{posicao} : {itens}")
```
