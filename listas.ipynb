```python
class No:
  def __init__(self, dado):
    self.dado = dado
    self.proximo = None
```

```python
no1 = No('Rafael')
no2 = No('Maria')
no3 = No('Otavio')
```

```python
no1.proximo = no2
no2.proximo = no3
```

```python
atual = no1

while atual is not None:
  print(atual.dado)
  atual = atual.proximo
```

```python
class Lista:
  def __init__(self):
    self.inicio = None

  def inserir(self, dado):
    novo = No(dado)

    if self.inicio is None:
      self.inicio = novo
    else:
      atual = self.inicio

      while atual.proximo is not None:
        atual = atual.proximo

      atual.proximo = novo

  def exibir(self):
    atual = self.inicio

    while atual is not None:
      print(atual.dado)
      atual = atual.proximo
```

```python
lista = Lista()
lista.inserir("Rodrigo")
lista.inserir("Thiago")
lista.inserir("Jéssica")

lista.exibir()
```
