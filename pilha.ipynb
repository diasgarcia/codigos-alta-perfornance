```python
class Pilha:
  def __init__(self):
    self.itens = []

  def empilhar(self, item):
    self.itens.append(item)

  def desempilhar(self):
    return self.itens.pop()

  def exibir(self):
    print(self.itens)
```

```python
nomes = Pilha()

nomes.empilhar("José")
nomes.empilhar("Maria")
nomes.empilhar("Ana")
nomes.empilhar("Carlos")

nomes.exibir()
```

```python
print(nomes.desempilhar())

nomes.exibir()
```
