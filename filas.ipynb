```python
class Fila:
  def __init__(self):
    self.itens = []

  def adicionar(self, item):
    self.itens.append(item)
    print(f"{item} entrou na fila!")

  def remover(self):
    if len(self.itens) == 0:
      print("A fila está vazia!")
      return None

    item = self.itens.pop(0)
    print(f"{item} foi removido!")
    return item

  def exibir(self):
    print(self.itens)
```

```python
fila = Fila()

fila.adicionar("Estudar estruturas de dados")
fila.adicionar("Estudar Java")
fila.adicionar("Fazer as unidades de aprendizagem")
```

```python
fila.exibir()
```

```python
fila.remover()
```
